```python
import os
import shutil

# The previous output showed the file is 'traffic.avi'
output_dir = 'runs/detect/predict-2'
src_path = os.path.join(output_dir, 'traffic.avi')
dest_path = "SmartSight-Detection-Framework/outputs/traffic_detected.mp4"

if os.path.exists(src_path):
    shutil.move(src_path, dest_path)
    print(f"Processed video saved successfully to: {dest_path}")
else:
    print(f"Error: {src_path} not found. Available files: {os.listdir(output_dir)}")
```
